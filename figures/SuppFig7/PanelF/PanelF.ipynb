# Supplemental Figure 7, panel F

Violin plot of nascent transcription level (PRO-seq log2(CPM+1)) at cREs that gain transcriptional activity with IFN-gamma stimulus in all 4 conditions (Vehicle, IFN-gamma, BAFinh, BAFinh+IFN-gamma)

*Outputs: PanelF.pdf*

# Setup

```r
library(tidyverse)
library(here)
here::i_am("figures/SuppFig7/PanelF/PanelF.ipynb")
sessionInfo()
```

# Load data

```r
#PRO-seq counts at GM12878 consensus cREs
cRE_counts <- read.delim(here("inputs","countmat","PROseq_eRNA","PROseq_cRE_counts.tsv"), sep='\t')
#DESeq2 results for IFNg vs Veh PRO-seq at cREs
ifn <- read.delim(here("inputs","countmat","PROseq_eRNA","IFN_v_Veh_PROseq_cRE.tsv"), sep='\t')

#convert counts to CPM
counts_only <- cRE_counts[,-1:-5]
cpm <- apply(counts_only, 2, function(x) 1000000*x/sum(x))
```

# Generate plot

```r
# generate object for plotting - mean CPM for all cREs with IFN-induced nascent transcription, in each conditioj
veh_cpm <- rowMeans(cpm[which(ifn$change=="up"),grepl("veh_veh", colnames(cpm))])
ifn_cpm <- rowMeans(cpm[which(ifn$change=="up"),grepl("veh_IFN", colnames(cpm))])
bafinh_cpm <- rowMeans(cpm[which(ifn$change=="up"),grepl("BAFinh_veh", colnames(cpm))])
bafinh_ifn_cpm <- rowMeans(cpm[which(ifn$change=="up"),grepl("BAFinh_IFN", colnames(cpm))])
cpm2 <- data.frame(veh_cpm, ifn_cpm, bafinh_cpm, bafinh_ifn_cpm)

# generate plot
cpm2 %>%
    pivot_longer(everything()) %>% 
    ggplot(aes(x=factor(name, levels=c("veh_cpm", "ifn_cpm", "bafinh_cpm", "bafinh_ifn_cpm")),y=log2(value+1))) +
    geom_violin() + geom_boxplot(width=.2, outlier.shape=NA) +
    theme_bw() + theme(panel.grid=element_blank(), text=element_text(size=20)) +
    xlab("") + ylab("log2(PRO-seq CPM +1)") +
    ylim(0,10)
ggsave("PanelF.pdf")

#Stats
#IFN vs Vehicle
wilcox.test(cpm2$ifn_cpm, cpm2$veh_cpm)$p.value
#IFN vs BAFinh+IFN
wilcox.test(cpm2$ifn_cpm, cpm2$bafinh_ifn_cpm)$p.value
#IFN vs BAFinh
wilcox.test(cpm2$ifn_cpm, cpm2$bafinh_cpm )$p.value
```
