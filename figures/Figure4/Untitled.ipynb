```r
library(tidyverse)
library(rGREAT)
```

```r
DAR <- read.delim("~/manuscripts/BAF_stimulus_CellGenomics/inputs/DARs/time_0.5h_DARs.tsv",sep='\t') %>% filter(TSS==FALSE)
background <- makeGRangesFromDataFrame(DAR)
```

```r
great_list <- list()
thresholds <- seq(2,4,by=.5)
for(i in 1:5){
    threshold <- thresholds[i]
    temp_gr <- makeGRangesFromDataFrame(DAR %>% filter(FDR < .05 & logFC < -log2(threshold)))
    great_res <- great(temp_gr, 
                       "msigdb:C5:GO:BP", 
                       "TxDb.Hsapiens.UCSC.hg38.knownGene",
                       background=background)
    great_list[[i]] <- getEnrichmentTable(great_res) %>% arrange(id)
}
```
