# Supplemental Figure 5, Panel A

PCA on PRO-seq gene-level counts, with first two PCs plotted.

*Output: PanelA.pdf*

# Setup

```r
library(tidyverse)
library(RColorBrewer)
library(DESeq2)
library(here)
here::i_am("figures/SuppFig5/PanelA/PanelA_PCA.ipynb")
sessionInfo()
```

# Load data

```r
dds <- readRDS(
        here("inputs", "DEGs", "GM12878", "BAFinh_stimulus", "PROseq", "PROseq_gene_dds.rds"))
```

# Generate plot

```r
plotPCA(rlog(dds), ntop=12437) + 
    geom_point(size=4) + 
    geom_vline(xintercept=0,linetype=2) + geom_hline(yintercept=0, linetype=2) +
    scale_color_brewer(palette="Paired") + theme_bw() +
    xlim(-15,15) + ylim(-10,10) + coord_fixed(ratio=30/20) + theme(text=element_text(size=20), panel.grid=element_blank()) +
    ggtitle("n = 12437")
ggsave("PanelA.pdf")
```
