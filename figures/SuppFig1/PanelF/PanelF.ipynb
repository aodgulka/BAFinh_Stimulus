# Supplemental Figure 1, Panel F

Adapted from Ernst et al., 2011, Figure 1B,C (https://www.nature.com/articles/nature09906)

*Output: state_features.pdf (histone mod/TF coverage by state), state_TSS_freq.pdf (% of each state that overlaps TSS), state_coverage.pdf (% of genome for each state)*

# Setup

```r
# load packages
library(tidyverse)
library(pheatmap)
library(RColorBrewer)
library(here)
here::i_am("figures/SuppFig1/PanelF/PanelF.ipynb")

sessionInfo()
```

# Load data

```r
#read in data
#values from Ernst et al., 2011, Figure 1B,C (https://www.nature.com/articles/nature09906)

states <- read.delim("Ernst_2011_15state_coverage.txt", sep='\t')
head(states)
```

# Generate plots

```r
#generate heatmaps for relevant states (enhancer, promoter, insulator):

#histone modification frequency by state
states %>% select(!TSS_2kb) %>% filter(!grepl("Heterochrom|Repetitive|Transcri", State, ignore.case=TRUE)) %>% column_to_rownames("State_num") %>% 
  select(!c(WCE,Median_coverage, State)) %>% 
  pheatmap(cluster_rows=F, cluster_cols=F, 
           color=colorRampPalette(colors=c("white", brewer.pal(6, "Paired")[6]))(100), 
           border_color=NA, 
           labels_row = states[-grep("Heterochrom|Repetitive|Transcri", states$State, ignore.case=TRUE),]$State, 
           cellwidth=15, cellheight=15, 
           breaks=seq(0,100, length.out=100),
           filename="state_features.pdf")

#TSS% by state
states %>% select(TSS_2kb, State, State_num) %>% filter(!grepl("Heterochrom|Repetitive|Transcri", State, ignore.case=TRUE)) %>% column_to_rownames("State_num") %>% 
  select(!State) %>% 
  pheatmap(cluster_rows=F, cluster_cols=F, 
           color=colorRampPalette(colors=c("white", brewer.pal(3, "Paired")[2]))(100), 
           border_color=NA, 
           labels_row = states[-grep("Heterochrom|Repetitive|Transcri", states$State, ignore.case=TRUE),]$State, 
           cellwidth=15, cellheight=15, 
           breaks=seq(0,100, length.out=100),
           filename="state_TSS_freq.pdf")

#genome coverage by state
states %>% filter(!grepl("TSS|Heterochrom|Repetitive|Transcri", State, ignore.case=TRUE)) %>% column_to_rownames("State_num") %>% 
  select(Median_coverage) %>% 
  pheatmap(cluster_rows=F, cluster_cols=F, 
           color=colorRampPalette(colors=c("white", "black"))(100), 
           border_color=NA, 
           labels_row = states[-grep("Heterochrom|Repetitive|Transcri", states$State, ignore.case=TRUE),]$State, 
           cellwidth=15, cellheight=15,
           filename="state_coverage.pdf")
```
