# Supplemental Figure 1, Panel E

Scatter plot of log2FC accessibilities for BRM014/Vehicle vs ACBI1/Vehicle. 

*Output: PanelE.pdf*

# Setup

```r
# load packages
library(tidyverse)
library(ggrastr)
library(here)
here::i_am("figures/SuppFig1/PanelE/PanelE.ipynb")

theme_set(theme_bw() + theme(panel.grid=element_blank(), text=element_text(size=15)))
```

# Load data

```r
#differential accessibility, ACBI1 (1uM) vs vehicle
ACBI1 <- read.delim(
  here("inputs", "DARs", "acbi1_DARs.tsv"),
  sep = "\t"
) %>%
  arrange(seqnames, start)

#Differential accessibility, BRM014 (10µM) vs vehicle
BRM014 <- read.delim(
  here("inputs", "DARs", "time_24h_DARs.tsv"),
  sep = "\t"
) %>%
  arrange(seqnames, start)
```

# Generate plot

```r
ggplot(data=NULL, aes(x=ACBI1$logFC, y=BRM014$logFC)) +
    geom_vline(xintercept=0, linetype=2) + geom_hline(yintercept=0, linetype=2) + 
    rasterize(geom_point(alpha=.1, size=.5), dpi=300) +
    geom_smooth(method=lm) +
    xlab("Log2 fold change accessibility\nACBI1 vs vehicle") +
    ylab("Log2 fold change accessibility\nBRM014 vs vehicle") +
    scale_x_continuous(breaks=c(-8,-4,0,4)) +
    scale_y_continuous(breaks=c(-8,-4,0,4)) +
    coord_cartesian(xlim=c(-9,4), ylim=c(-9,4))

ggsave("PanelE.pdf", width=4, height=4)
```
