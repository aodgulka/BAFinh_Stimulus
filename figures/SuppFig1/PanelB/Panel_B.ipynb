# Supplementary Figure 1, Panel B

Comparison of accessibility log2FC between BAFinh timepoints (at 10uM) and doses (at 6h). Bottom half of each plot grid contains pairwise scatter plots, while top half contains heatmap of Pearson r values

*Outputs: panel_b_heatmap_scatter_time.pdf, panel_b_heatmap_scatter_dose.pdf*

# Setup

```r
# load packages
library(tidyverse)
library(patchwork)
library(ggrastr)
library(here)
here::i_am("figures/SuppFig1/PanelB/Panel_B.ipynb")

sessionInfo()

#function to calculate Pearson correlation between logFCs for two BAFinh conditions
get_r <- function(x, y) {
    r <- cor(logFCs[[x]], logFCs[[y]], use = "complete.obs")
    r
}

# function to make scatterplot of logFCs between two BAFinh conditions
make_scatter <- function(x, y, name_x, name_y) {
    
    ggplot(logFCs, aes(x = .data[[x]], y = .data[[y]])) +
        #geom_abline(linetype=2) + #parity line (ie what we would see with a perfect correlation)
        rasterize(geom_point(alpha = 0.1, size = .05, color = "black"), dpi=300) + 
        geom_smooth(method = "lm", se = FALSE, color = "firebrick", linewidth = .5) + #plot linear best fit
        theme_bw(base_size = 10) +
        labs(x = name_x, y = name_y) +
        theme(
            axis.text = element_text(color = "black", size = 5),
            axis.title = element_text(size = 5),
            axis.ticks = element_line(color = "black", linewidth = .5),
            plot.title = element_text(hjust = 0.5), 
            panel.grid=element_blank()
        ) +
        scale_x_continuous(limits=c(-8,2), breaks=c(-8,-4,0)) +
        scale_y_continuous(limits=c(-8,2), breaks=c(-8,-4,0)) 
}


#for each pairwise comparison, generate heatmap indicating strength of correlation (Pearson's r)
make_heatmap <- function(r) {
  ggplot(data.frame(x=1, y=1, pval=r), aes(x, y, fill=r)) +
    geom_tile(color = "black", size = 0.2) +
    scale_fill_gradient(low="white", high="firebrick", limits=c(0,1)) +
    geom_text(aes(label=sprintf("%.2f",r)), size=4, fontface="bold") +
    theme_void() +
    theme(panel.border = element_rect(color="black", fill=NA, size=0.3))
}
```

# Load data

```r
#get list of files for all DAR analyses
files <- list.files(
    here("inputs", "DARs"),
    pattern    = "*.tsv",
    full.names = TRUE
)

#generate single data frame containing all conditions
df <- map_dfr(files, 
              function(f) {
                tmp <- read_tsv(f, col_names=TRUE)
                label <- tools::file_path_sans_ext(basename(f))
                tmp$label <- label #add label column indicating treatment condition
                tmp
                }
             )
#separate into time and dose data frames
df_time <- df %>% select(ID, logFC, label) %>% filter(grepl("time", label))
df_dose <- df %>% select(ID, logFC, label) %>% filter(grepl("dose", label)) 

#pivot to wide format, where every column contains log2 accessibility fold changes for one BAFinh treatment condition
logFCs <- df %>% pivot_wider(id_cols=ID,names_from="label", values_from="logFC") 
```

# Generate plots

```r
#timecourse plot

plots <- list()
labels <- unique(df_time$label)[c(1,2,4,5,3)] #reorder labels to be first in order of ascending time at 10uM (0.5h >>> 24h)

#generate scatter plots and heatmap tiles for all pairwise comparisons
for (i in seq_along(labels)) {
  row_plots <- list()
  for (j in seq_along(labels)) {
    if (i == j) {
      p <- ggplot() + theme_void() #for diagonal (self-correlation) we don't need plots
    } else if (i > j) {
      p <- make_scatter(labels[j], labels[i], labels[j], labels[i]) #lower half of panel - scatter plot of logFCs
    } else {
      r <- get_r(labels[j], labels[i]) #upper half of panel - heatmap of r values
      p <- make_heatmap(r)
    }
    row_plots[[j]] <- p
  }
  plots[[i]] <- row_plots
}

final_plot_time <- wrap_plots(plotlist = unlist(plots, recursive = FALSE), ncol = 5) + 
    plot_layout(guides = "collect") & theme(legend.position = "right") #specify  position for heatmap colorbar

ggsave("panel_b_heatmap_scatter_time.pdf", final_plot_time, width = 5.5, height = 5, useDingbats=FALSE)
```

```r
#dosage plot

plots <- list()
labels <- unique(df_dose$label)[c(2,1,4,3)] #reorder labels to be first in ascending order of dose (10nM, 100nM, 1uM) 

#generate scatter plots and heatmap tiles for all pairwise comparisons
for (i in seq_along(labels)) {
  row_plots <- list()
  for (j in seq_along(labels)) {
    if (i == j) {
      p <- ggplot() + theme_void() #for diagonal (self-correlation) we don't need plots
    } else if (i > j) {
      p <- make_scatter(labels[j], labels[i], labels[j], labels[i]) #lower half of panel - scatter plot of logFCs
    } else {
      r <- get_r(labels[j], labels[i]) #upper half of panel - heatmap of r values
      p <- make_heatmap(r)
    }
    row_plots[[j]] <- p
  }
  plots[[i]] <- row_plots
}

final_plot_time <- wrap_plots(plotlist = unlist(plots, recursive = FALSE), ncol = 4) + 
    plot_layout(guides = "collect") & theme(legend.position = "right") #specify  position for heatmap colorbar

ggsave("panel_b_heatmap_scatter_dose.pdf", final_plot_time, width = 4.5, height = 4, useDingbats=FALSE)
```
