# Supplemental Figure 2, Panel G

Heatmap of ridge regression coefficients (left) and MDI feature importance (right) from models trained on different BAFinh doses. Heatmaps are clustered based on column Z-scored ridge coefficients. 

*Output: PanelG.pdf*

# Setup

```r
library(tidyverse)
library(pheatmap)
library(RColorBrewer)
theme_set(theme_bw() + theme(panel.grid=element_blank()))
library(gridExtra)
library(here)
here::i_am("figures/SuppFig2/PanelG/PanelG.ipynb")

sessionInfo()
```

# Load data

```r
#read in feature importances for ReModEx run on different DAR sets (varying BAFinh dose)


dat_10nM <- read.delim(
  here(
    "inputs", "ReModEx", "GM12878", "all_cRE",
    "dose_10nM", "feature_importance.csv"
  ),
  sep = ","
) %>%
  arrange(feature)

dat_100nM <- read.delim(
  here(
    "inputs", "ReModEx", "GM12878", "all_cRE",
    "dose_100nM", "feature_importance.csv"
  ),
  sep = ","
) %>%
  arrange(feature)

dat_1uM <- read.delim(
  here(
    "inputs", "ReModEx", "GM12878", "all_cRE",
    "dose_1uM", "feature_importance.csv"
  ),
  sep = ","
) %>%
  arrange(feature)

dat_10uM <- read.delim(
  here(
    "inputs", "ReModEx", "GM12878", "all_cRE",
    "dose_10uM", "feature_importance.csv"
  ),
  sep = ","
) %>%
  arrange(feature)
```

# Generate plots

```r
#compare Ridge coefficients across doses
#hierarchical clustering based off of coeffs (column Z-score normalized to mitigate effects of different range of scores between timepoints)
Ridge_coef_df <- data.frame(row.names=dat_10nM$feature, "10nM"=dat_10nM$Ridge_coef, "100nM"=dat_100nM$Ridge_coef, "1uM"=dat_1uM$Ridge_coef, "10uM"=dat_10uM$Ridge_coef)
MDI_df <- data.frame(row.names=dat_10nM$feature, "10nM"=dat_10nM$MDI_importance, "100nM"=dat_100nM$MDI_importance, "1uM"=dat_1uM$MDI_importance, "10uM"=dat_10uM$MDI_importance)

dist_ridge <- dist(scale(Ridge_coef_df))
hclust_ridge <- hclust(dist_ridge)

p1 <- pheatmap((Ridge_coef_df[hclust_ridge$order,]), cluster_rows=FALSE, cluster_cols=FALSE,
        color=colorRampPalette(colors=c(brewer.pal(3, "Paired")[2], "white", brewer.pal(6, "Paired")[6]))(100), fontsize_row=3,
        main="Cluster column Z-scored ridge coef\nPlot raw ridge coef", width=3,
              breaks=c(seq(-.2, 0, length.out=50)[1:49], seq(0, .2, length.out=51)), silent=TRUE)

p2 <- pheatmap((MDI_df[hclust_ridge$order,]), cluster_rows=FALSE, cluster_cols=FALSE,
        color=colorRampPalette(colors=c("white", "black"))(100), fontsize_row=3,
        main="Cluster column Z-scored ridge coef\nPlot raw MDI", width=3, silent=TRUE, breaks=seq(0,0.1,length.out=100))

```

# Assemble figure panel

```r
pdf("PanelG.pdf")
gridExtra::grid.arrange(p1$gtable,p2$gtable, ncol=2)
dev.off()
```
