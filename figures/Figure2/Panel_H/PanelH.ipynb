# Figure 2, Panel H

Heatmap of ridge coefficients for 46 features shared between GM12878, HepG2, and H1 ChIP-seq data series, ordered by mean coefficient across all cell types

*Output: PanelH.pdf*

# Setup

```r
# load packages
library(tidyverse)
library(pheatmap)
library(RColorBrewer)
library(here)
here::i_am("figures/Figure2/Panel_H/PanelH.ipynb")

sessionInfo()
```

# Load data (feature importance metrics from GM12878, HepG2, H1)

```r
# load feature importance data (from enh model run only on shared features)
#GM12878
gm <- read.delim(
  here(
    "inputs", "ReModEx", "all_cells_features",
    "GM12878", "feature_importance.csv"
  ),
  sep = ","
) %>%
  select(feature, Ridge_coef)

#HepG2
hepg2 <- read.delim(
  here(
    "inputs", "ReModEx", "all_cells_features",
    "HepG2", "feature_importance.csv"
  ),
  sep = ","
) %>%
  select(feature, Ridge_coef)

#H1 ESC
h1 <- read.delim(
  here(
    "inputs", "ReModEx", "all_cells_features",
    "H1", "feature_importance.csv"
  ),
  sep = ","
) %>%
  select(feature, Ridge_coef)
```

# Generate figure panel

```r
#consolidate ridge coefs across all cell types
all <- inner_join(inner_join(gm, hepg2, by="feature"), h1, by = "feature") %>% `colnames<-`(c("feature", "GM", "HepG2", "H1")) %>% column_to_rownames("feature")

#plot, ordering rows by mean Ridge coef
heatmap <- pheatmap(all[order(rowMeans(all)),], 
                    color=colorRampPalette(colors=c(brewer.pal(3,"Paired")[2], "white", brewer.pal(6, "Paired")[6]))(100), 
                    breaks=c(seq(-0.2,0, length.out=51)[-51], seq(0, 0.2, length.out=50)), 
                    border_color=NA, 
                    cluster_rows=F, 
                    cluster_cols=F, 
                    filename="PanelH.pdf", width=4)
```
